```python
##Fecha	Cliente ID	Producto	Categoría	Cantidad	Precio Unitario	Costo Unitario	Sede
##3/08/2026	101	Bowl Andino	Almuerzo	2	24.000,00	14.000,00	centro
##3/08/2026	102	Pasta Urbana	Almuerzo	1	28.000,00	17.000,00	NORTE
##4/08/2026	103	Bowl Andino	Almuerzo	1	24.000,00	14.000,00	Centro


datos_ventas = {"Fecha": ["3/08/2026", "3/08/2026", "4/08/2026"],
  "Cliente_ID": [101, 102, 103],
  "Producto": ["Bowl", "Pasta", "Bowl"],
  "Categoría": ["Almuerzo", "Almuerzo", "Almuerzo"],
"Cantidad": [2, 1, 1],
"Precio_Unitario": [24000, 28000, 24000],
"Costo_Unitario": [14.000, 17.000, 14.000],
"Sede": ["centro", "NORTE", "Centro"]}

print(datos_ventas)
```

```python
import pandas as pd

df_ventas = pd.DataFrame(datos_ventas)
type(df_ventas)

df_ventas.head()
```

```python
# una columna -> Series
df_ventas["Producto"]

# varias columnas -> DataFrame
columnas = ["Fecha", "Producto", "Cantidad", "Sede"]
df_ventas[columnas]

# columnas por posición
df_ventas.iloc[:, 0:3]
```

```python
mascara = (
(df_ventas["Sede"] == "Centro")
& (df_ventas["Categoría"] == "Almuerzo")
)

ventas_centro_almuerzo = df_ventas.loc[mascara, columnas].copy()

unidades_centro_almuerzo = ventas_centro_almuerzo["Cantidad"].sum()
```

```python
df_ventas["ingreso"] = df_ventas["Cantidad"] * df_ventas["Precio_Unitario"]
df_ventas["costo_total"] = df_ventas["Cantidad"] * df_ventas["Costo_Unitario"]
df_ventas["margen_bruto"] = df_ventas["ingreso"] - df_ventas["costo_total"]
df_ventas["porcentaje_margen"] = (
df_ventas["margen_bruto"] / df_ventas["ingreso"] * 100

)

print(df_ventas)
```

```python
kpi_producto = (
df_ventas
.groupby("Producto")
.agg(
unidades=("Cantidad", "sum"),
ingreso=("ingreso", "sum"),
margen=("margen_bruto", "sum"),
)
.reset_index()
)

kpi_producto["porcentaje_margen"] = (
kpi_producto["margen"] / kpi_producto["ingreso"] * 100
).round(2)

print(kpi_producto)
```

```python
kpi_producto = kpi_producto.sort_values("ingreso", ascending=False)

producto_lider = kpi_producto.iloc[0]["Producto"] # Bowl Andino

# alternativa sin ordenar
indice = kpi_producto["ingreso"].idxmax()
producto_lider = kpi_producto.loc[indice, "Producto"]

print(producto_lider)
```
