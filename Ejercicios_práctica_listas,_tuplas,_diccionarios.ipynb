**28.** Agrupe los nombres de los clientes según su nivel de riesgo

```python
clientes = [
    {"nombre": "Ana", "riesgo": "bajo"},
    {"nombre": "Luis", "riesgo": "alto"},
    {"nombre": "Marta", "riesgo": "bajo"}
]
grupo = {}

for c in clientes:
  riesgo = c["riesgo"]
  if riesgo not in grupo:
    grupo[riesgo]= []
  grupo[riesgo].append(c["nombre"])

print(grupo)
```

**30. Mini caso integrador**

Complete el algoritmo que calcula el total de los créditos mayores o iguales a 100000 y registra los clientes correspondientes

```python
transacciones = [
    {"cliente": "Ana", "tipo": "Credito", "monto": 120000},
    {"cliente": "Luis", "tipo": "Debito", "monto": 85000},
    {"cliente": "Marta", "tipo": "Credito", "monto": 210000}
]
total_creditos = 0
clientes = []

for t in transacciones:
  if t["tipo"] == "Credito" and t["monto"] >= 100000:
    total_creditos += t["monto"]
    clientes.append(t["cliente"])

print(total_creditos)
print(clientes)
```

**Mi respuesta:** En el ejercicio aparecen listas, diccionarios, un ciclo for y una condición if. La lista transacciones almacena todas las operaciones y cada transacción es un diccionario que relaciona los datos cliente, tipo y monto. El ciclo for recorre cada transacción y el if selecciona únicamente aquellas que sean de tipo "Credito" y tengan un monto mayor o igual a 100000. Finalmente, += acumula el valor de los créditos y append() guarda los nombres de los clientes que cumplen las condiciones.
