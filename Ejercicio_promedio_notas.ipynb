```python
def promedio (notas):
  total = 0
  for n in notas:
    total = total + n
  return total / len(notas)

def aprueba(prom, minimo=3.0):
  return prom >= minimo

notas = [3.5, 4.2, 2.8]
prom = promedio(notas)

print("Promedio:", prom)
if aprueba(prom):
  print("Aprobado")
else:
  print("No aprobado")
```
